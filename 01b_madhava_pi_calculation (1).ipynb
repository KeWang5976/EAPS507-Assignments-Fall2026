# Madhava-Leibniz π calculation


This notebook is designed to be a live demonstration of Python fundamentals applied to calculating π using the [Madhava-Leibniz series](https://en.wikipedia.org/wiki/Approximations_of_%CF%80#Middle_Ages)

The approximation is simple:

$$ \pi \approx 4 \cdot \left(1 - \frac{1}{3} + \frac{1}{5} - \frac{1}{7} + \frac{1}{9} \ldots -(-1)^n \frac{1}{2n - 1}  \right) $$

```python
import math
import matplotlib.pyplot as plt

# number of iterations
N = 5000

# starting value
pi_value = 0

# empty lists to store results
iterations = []
pi_values = []
errors = []

# calculate pi
for n in range(1, N + 1):

    if n == 1:
        term = 1
    else:
        term = -((-1)**n) / (2*n - 1)

    pi_value = pi_value + term

    calculated_pi = 4 * pi_value
    error = abs(calculated_pi - math.pi)

    iterations.append(n)
    pi_values.append(calculated_pi)
    errors.append(error)

print("Calculated pi:", calculated_pi)
print("Expected pi:", math.pi)
print("Absolute error:", error)
```

## Part 1 

I used the Madhava-Leibniz series to approximate pi. I changed the original
example so that it calculates pi using different numbers of iterations.


For this assignment, I consider an absolute error below 0.001 to be
reasonably accurate.

```python
print("10 iterations:")
print("Calculated pi =", pi_values[9])
print("Absolute error =", errors[9])

print()

print("100 iterations:")
print("Calculated pi =", pi_values[99])
print("Absolute error =", errors[99])

print()

print("1000 iterations:")
print("Calculated pi =", pi_values[999])
print("Absolute error =", errors[999])

print()

print("4000 iterations:")
print("Calculated pi =", pi_values[3999])
print("Absolute error =", errors[3999])
```

```python
reasonable_error = 0.001

for i in range(len(errors)):
    if errors[i] < reasonable_error:
        print("Reasonable accuracy is reached at", iterations[i], "iterations.")
        print("Calculated pi:", pi_values[i])
        print("Absolute error:", errors[i])
        break
```

```python
plt.plot(iterations, errors)

plt.xlabel("Iterations")
plt.ylabel("Error")
plt.title("Pi Approximation Error")

plt.show()
```

### Result
As the number of iterations increased, the error became smaller. My
reasonable accuracy level was 0.001, and the calculation reached this level
at about 1000 iterations.

The graph shows that the calculated value gets closer to pi when more
iterations are used. 

## Part 2 - Function

```python
def calculate_pi(n):
    """
    Calculate an approximation using n iterations.
    """

    pi_value = 0

    for i in range(1, n + 1):

        if i == 1:
            term = 1
        else:
            term = -((-1)**i) / (2*i - 1)

        pi_value = pi_value + term

    return 4 * pi_value
```

```python
print("10 iterations:", calculate_pi(10))
print("100 iterations:", calculate_pi(100))
print("1000 iterations:", calculate_pi(1000))

print("Expected pi:", math.pi)
```
